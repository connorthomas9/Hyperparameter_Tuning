## Introduction

In this project, we will be using different hyperparameter tuning methods to classify the different types of raisins. This dataset has been posted on `Kaggle` by `Murat Koklu`, a researcher who has studied different raisin grain types using machine learning methods.

In this dataset, there are two types of raisin grains; `Kecimen` and `Besni` and seven numerical predictor variables associated with each of the 900 samples in the data.

The two hyperparameter tuning methods that will be used in this project will be:

1. Grid Search method to tune a Decision Tree Classifier
2. Random Search method to tune a Logistic Regression Classifier

## Data

The dataset contains 7 explanatory features as well as one additonal feature classifying the raisin as `Kecimen` or `Bensi`:

- `Area`: Area of the raisin grain.
- `MajorAxisLength`: Major axis length of the raisin grain.
- `MinorAxisLength`: Minor axis length of the raisin grain.
- `Eccentricity`: Eccentricity of the raisin.
- `ConvexArea`: Convex area of the raisin
- `Extent`: Extent of the raisin
- `Perimeter`: Perimeter of the raisin
- `Class`: Class of the raisin; either `Kecimen` or `Besni`

```python
# Importing necessary packages and modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
%matplotlib inline
```

```python
# Loading data file into Pandas DataFrame
raisins = pd.read_csv("Raisin_Dataset.csv")
```

## Analysis

The first thing that we should do now that we have installed and imported the necessary packages and modules, as well as loading our data file into a Pandas DataFrame, is we should just take a quick look at the first few rows of our dataset.

```python
# Displaying the first few rows of the data
print(raisins.head())
```

### Converting `Class` into Binary Categorization

Upon inspecting our dataset, all looks to be in the format we want, except for the `Class` column.

Ideally, we would like for this column to hold values of `0` for Kecimen and `1` for Besni.

Let's fix that really quick.

```python
# Converting class values to binary
raisins['Class'] = raisins['Class'].replace({'Kecimen': 0, 'Besni': 1})
```

Now, let's see if this change worked.

In adddition to this, we will also print the `dtypes` of the variables in order to make sure that all columns are of a numerical data type.

```python
# Displaying the first few rows of data
print(raisins.head())

# Displaying dtypes
print(raisins.dtypes)
```

Again, we can see that our `Class` column is not in the correct format that we need it to be in, so we will quickly chnage the dtype from `object` to `int64`.

```python
# Changing dtype of Class column to integer
raisins['Class'] = pd.to_numeric(raisins['Class'])

# Displaying column dtypes
print(raisins.dtypes)
```

### Creating Predictor Variables

Now that we have our data formatted correctly, we need to define our explanatory (`X`) and dependent variable (`y`).

```python
# Defining explanatory and dependent variables
X = raisins.drop('Class', axis=1)
y = raisins['Class']
```

### Examining our Dataset

Before we split our data into training and testing sets, we should just take a moment to see the actual size of our dataset.

```python
# Examining the dataset
print("Number of Features: ", X.shape[1])
print("Total Number of Samples: ", len(y))
print("Samples Belonging to Class '1': ", y.sum())
```

### Splitting our data into testing and training sets

Now that we have an overall idea of our dataset, we can split the data into training and testing sets.

```python
# Splitting the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 19)
```

## Grid Search with Decision Tree Classifier

Now with our data split into our training and testing sets, we are ready to begin creating our decison tree classifier.

```python
# Creating decision tree classifier model
tree = DecisionTreeClassifier()
```

### Dictionary of Parameters for GridSearchCV

Now, we will define the parameters for our future GridSearchCV model.

```python
# Defining dictionary of parameters for GridSearchCV model
parameters = {'min_samples_split': [2,3,4], 'max_depth': [3,5,7]}
```

### Creating and Fitting `GridSearchCV` Model

Next, we can simply define our GridSearchCV model, and then fit that same model.

```python
# Creating GridSearchCV model
grid = GridSearchCV(tree, parameters)

# Fitting the model to the training data
grid.fit(X_train, y_train)
```

### Model and Hyperparameters, Best Score, and Accuracy Obtained by GridSearchCV

Now that we have created and fitted our GridSearchCV model, we can now print the `max_depth` and `min_samples_split`, as well as the best score of the model, and the overall accuracy score of the final model when run on the test data.

```python
# Displaying the model and hyperparameters
print(grid.best_estimator_)

# Displaying best score
print("Best Score: ", grid.best_score_)

# Displaying accuracy of model on testing dataset
print("Accuracy Score: ", grid.score(X_test, y_test))
```

### GridSearchCV Results Summary

Lastly, we can display a summary table of the results of our GridSearchCV model.

```python
# Displaying a table of results summarizing the GridSearchCV model
df = pd.concat([pd.DataFrame(grid.cv_results_['params']), pd.DataFrame(grid.cv_results_['mean_test_score'], columns=['Score'])], axis=1)
print(df)
```

## Random Search with Logistic Regression

Now that we have our GridSearchCV model complete, we can now begin the process of our random search with a logistic regression model.

The first thing that we need to do is to define our logistic regression model.

```python
# Defining logistic regression model
lr = LogisticRegression(solver = 'liblinear', max_iter = 1000)
```

### Defining Distributions to choose hyper parameters from

```python
# Defining distributions
from scipy.stats import uniform
distributions = {'penalty': ['l1', 'l2'], 'C': uniform(loc=0, scale=100)}
```

### Defining and Fitting a `RandomizedSearchCV` Model

```python
# Defining RandomizedSearchCV model
clf = RandomizedSearchCV(lr, distributions, n_iter=8)

# Fitting the model to the training data
clf.fit(X_train, y_train)
```

### Best Estimator, Best Score, and Summary Table of Results for RandomSearchCV Model

```python
# Displaying best estimator and best score
print("Best Estimator: ", clf.best_estimator_)
print("Best Score: ", clf.best_score_)
```

```python
# Displaying summary table of results
df_2 = pd.concat([pd.DataFrame(clf.cv_results_['params']), pd.DataFrame(clf.cv_results_['mean_test_score'], columns=['Accuracy'])], axis=1)
print(df_2.sort_values('Accuracy', ascending = False))
```
