## Introduction

In this project, we will be using different hyperparameter tuning methods to classify the different types of raisins. This dataset has been posted on `Kaggle` by `Murat Koklu`, a researcher who has studied different raisin grain types using machine learning methods.

In this dataset, there are two types of raisin grains; `Kecimen` and `Besni` and seven numerical predictor variables associated with each of the 900 samples in the data.

The two hyperparameter tuning methods that will be used in this project will be:

1. Grid Search method to tune a Decision Tree Classifier
2. Random Search method to tune a Logistic Regression Classifier

## Data

The dataset contains 7 explanatory features as well as one additonal feature classifying the raisin as `Kecimen` or `Bensi`:

- `Area`: Area of the raisin grain.
- `MajorAxisLength`: Major axis length of the raisin grain.
- `MinorAxisLength`: Minor axis length of the raisin grain.
- `Eccentricity`: Eccentricity of the raisin.
- `ConvexArea`: Convex area of the raisin
- `Extent`: Extent of the raisin
- `Perimeter`: Perimeter of the raisin
- `Class`: Class of the raisin; either `Kecimen` or `Bensi`
